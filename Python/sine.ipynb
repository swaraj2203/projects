```python
a = np.linspace(-10*np.pi, 10*np.pi, 200)
b = np.linspace(-10*np.pi, 10*np.pi, 200)
```

```python
A, B = np.meshgrid(a,b)
```

```python
r = np.sqrt(A**2 + B**2)
```

```python
R = np.sin(r)
```

```python
plt.imshow(
    R,
    extent=[a.min(), a.max(), b.min(), b.max()],
    origin='lower'
)
plt.colorbar()
plt.xlabel('a')
plt.ylabel('b')
plt.show()
```
