```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x_vals = np.linspace(-4*(np.pi),4*np.pi,10000)
```

```python
def sq_wave(x):
    while x > np.pi: 
        x = x - (2 * np.pi)
    while x < -np.pi:
        x = x + (2 * np.pi)
        
    if x < 0:
        return -(1)
    else:
        return 1
```

While loop ensures the periodicity \
when x is less/greater than $\pm\pi$, the value assigned to it should satisfy \
f(x + T) = f(x)

```python
y_vals = np.array([sq_wave(x) for x in x_vals]) 
```

```python
def f_sq(x_vals):

    total = np.zeros_like(x_vals)

    n = np.arange(0,20)
    m = 2*n + 1

    for k in m:
        total += (4/(k*np.pi))*np.sin(k*x_vals)

    return total
```

```python
y_f = f_sq(x_vals)
```

```python
fig, ax = plt.subplots(figsize=(10,5))

ax.spines["left"].set_position("zero")
ax.spines["bottom"].set_position("zero")
ax.spines["right"].set_color("none")
ax.spines["top"].set_color("none")

ticks = [-3*np.pi, -2*np.pi, -np.pi, 0, np.pi, 2*np.pi, 3*np.pi]
labels = [r"$-3\pi$", r"$-2\pi$", r"$-\pi$", "0", r"$\pi$", r"$2\pi$", r"$3\pi$"]

ax.set_xlim(-3*np.pi, 3*np.pi)
ax.set_xticks(ticks)
ax.set_xticklabels(labels)
ax.set_ylim(-1.5, 1.5)

ax.plot(x_vals,y_vals,label='Square wave')
ax.plot(x_vals,y_f,label='Fourier Series')
plt.legend()
plt.show()
```

```python
def saw_wave(x):
    while x > np.pi:
        x = x - 2*np.pi
    while x < -np.pi:
        x = x + 2*np.pi

    return x
```

```python
y_saw = np.array([saw_wave(x) for x in x_vals])
```

```python
def f_sw(x_vals):

    total = np.zeros_like(x_vals)

    n = np.arange(1,40)

    for k in n:
        total += 2/k * (-1)**(k+1) * np.sin(k*x_vals)

    return total
```

```python
y_fsaw = np.array([f_sw(x) for x in x_vals])
```

```python
fig, ax = plt.subplots(figsize=(10,5))

ax.spines["left"].set_position("zero")
ax.spines["bottom"].set_position("zero")
ax.spines["right"].set_color("none")
ax.spines["top"].set_color("none")

ticks = [-3*np.pi, -2*np.pi, -np.pi, 0, np.pi, 2*np.pi, 3*np.pi]
labels = [r"$-3\pi$", r"$-2\pi$", r"$-\pi$", "0", r"$\pi$", r"$2\pi$", r"$3\pi$"]

ax.set_xlim(-3*np.pi, 3*np.pi)
ax.set_xticks(ticks)
ax.set_xticklabels(labels)
ax.set_ylim(-5, 5)

ax.plot(x_vals,y_saw,label='Sawtooth wave')
ax.plot(x_vals, y_fsaw, label='Fourier Series')
plt.legend()
plt.show()
```

```python
def tri_wave(x):
    while x < -np.pi:
        x = x + 2*np.pi
    while x > np.pi:
        x = x - 2*np.pi
        
    return 1 - 2/np.pi * abs(x)
```

```python
y_t = np.array([tri_wave(x) for x in x_vals])
```

```python
def f_tri(x):

    total = np.zeros_like(x)
    
    n = np.arange(1,10,2)

    for k in n:
        total += 8/(k**2 * (np.pi)**2) * np.cos(k*x)

    return total
```

```python
y_ftri = np.array([f_tri(x) for x in x_vals])
```

```python
fig, ax = plt.subplots(figsize=(10,5))

ax.spines["left"].set_position("zero")
ax.spines["bottom"].set_position("zero")
ax.spines["right"].set_color("none")
ax.spines["top"].set_color("none")

ticks = [-3*np.pi, -2*np.pi, -np.pi, 0, np.pi, 2*np.pi, 3*np.pi]
labels = [r"$-3\pi$", r"$-2\pi$", r"$-\pi$", "0", r"$\pi$", r"$2\pi$", r"$3\pi$"]

ax.set_xlim(-3*np.pi, 3*np.pi)
ax.set_xticks(ticks)
ax.set_xticklabels(labels)
ax.set_ylim(-5, 5)

ax.plot(x_vals,y_t,label='Triangle wave')
ax.plot(x_vals, y_ftri, label='Fourier Series')
plt.legend()
plt.show()
```
